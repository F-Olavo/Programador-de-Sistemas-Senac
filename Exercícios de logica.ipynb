# Exercícios de Programação


1. Faça um algoritmo para ler dois números e imprimir a soma dos números lidos.



```python
numero1 = float (input('Informe o 1° número: '))
numero2 = float (input('Informe o 2° número: '))
#print(type(numero1))
soma = numero1 + numero2
print(f'O resultado è: {soma}')
```

# Exercícios de Programação


2. Faça um algoritmo para ler três números e imprimir a soma, média e produto dos números lidos.




```python
numero1 = float (input('Informe o 1° número: '))
numero2 = float (input('Informe o 2° número: '))
numero3 = float (input('Informe o 3° número: '))
soma = numero1 + numero2 + numero3
media = soma / 3
produto = numero1 * numero2 * numero3
print(f'O resultado da soma è: {soma}')
print(f'O resultado da média è: {media}')
print(f'O resultado da produto è: {produto}')
```

# Exercícios de Programação


3. Faça um algoritmo para ler dois números e imprimir o maior, o menor ou então dizer se são iguais.


```python
numero1 = float(input('Informe o 1° número: '))
numero2 = float(input('Informe o 2° número: '))

if numero1 > numero2:
    print(f'O maior número é: {numero1}')
    print(f'O menor número é: {numero2}')
elif numero1 < numero2:
    print(f'O maior número é: {numero2}')
    print(f'O menor número é: {numero1}')
else:
    print('Os números são iguais')
```

# Exercícios de Programação


4. Faça um algoritmo para ler um número inteiro e dizer se o número lido é par ou impar.


```python
numero = int (input('Informe número: '))
print(type(numero))
result = numero % 2
if result == 0:
    print(f'O número {numero} é par')
else:
    print(f'O número {numero} é impar')
```

5. Faça um algoritmo para ler dois números A e B e dizer se A é divisível por B.

```python
numero1 = int(input('Informe o 1° número: '))
numero2 = int(input('Informe o 2° número: '))
result = numero1 % numero2
if result == 0:
  print(f'O {numero1} è divisível por {numero2}.')
else:
  print(f'O {numero1} não è divisível por {numero2}.')
```
